```python
import numpy as np
import csv

print("Finished importing libraries.")
```

```python
##### DEFINE CONSTANTS #####

# used in output
outputFilename = "z_AllCombResults/SmallStrideSearch/E1_D1_THO_Extended.csv"


# Select Files to Combine
DatasetSelect = 5

if DatasetSelect == 1:
	FilesToRead = [
		'z_AllCombResults/z_CompResults_OHO_A1332_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_OHO_A1333_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_OHO_A1342_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_OHO_A1343_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_OHO_A1352_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_OHO_A1353_E1_D1_T99.txt'
	]
	print("Using OHO Data")
elif DatasetSelect == 2:
	FilesToRead = [
		# 'z_AllCombResults/z_CompResults_THO_A1332_E1_D1_T99.txt',
		# 'z_AllCombResults/z_CompResults_THO_A1333_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_THO_A1342_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_THO_A1343_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_THO_A1352_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_THO_A1353_E1_D1_T99.txt'
	]
	print("Using THO Data")

elif DatasetSelect == 3:
	FilesToRead = [
		'z_AllCombResults/z_CompResults_Clem_A1332_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_Clem_A1333_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_Clem_A1342_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_Clem_A1343_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_Clem_A1352_E1_D1_T99.txt',
		'z_AllCombResults/z_CompResults_Clem_A1353_E1_D1_T99.txt'
	]
	print("Using Clemson Data")
elif DatasetSelect == 4:
	outputFilename = "z_AllCombResults/zz_E2_D1_T12_DomOHO.csv"
	FilesToRead = [
		'z_AllCombResults/z_CompResults_DOHOv2_A1333_WS05_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DOHOv2_A1333_WS10_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DOHOv2_A1333_WS15_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DOHOv2_A1333_WS20_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DOHOv2_A1333_WS25_E2_D1_T12.txt',
	]
	print("Using Dom OHO Data, Pt2Window Eval (12 sec tol)")

elif DatasetSelect == 5:
	outputFilename = "z_AllCombResults/zz_E2_D1_T12_DomClem.csv"
	FilesToRead = [
		'z_AllCombResults/z_CompResults_DCClem_A1333_WS05_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DCClem_A1333_WS10_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DCClem_A1333_WS15_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DCClem_A1333_WS20_E2_D1_T12.txt',
		'z_AllCombResults/z_CompResults_DCClem_A1333_WS25_E2_D1_T12.txt',
	]
	print("Using DomClem Data, Pt2Window Eval (12 sec tol)")


# end of dataset select


# Used in Sorting Data
F1_idx = 0
Fold_idx = 7
numFolds = 5




print("Finished Defining Constants")
```

```python
### Read in each file and add to output array



AllFileData = []

for currFilename in FilesToRead:
	with open(currFilename,'r') as fp:
		print("Reading file {} ...".format(currFilename))

		currFileData = []
		for row in fp:
			row = row.split(' ')
			row[-1] = row[-1].split('\n')[0] # remove final new line char
			currFileData.append(row)
		# end of for row
	# end with open()

	# Sort all into groups of 5 fold, but within each fold ranked highest to lowest F1

	sortedData = sorted(currFileData, key=lambda x: (x[Fold_idx], -float(x[F1_idx]))) # negative F1 to get descending order

	numSets = int(len(sortedData)/numFolds)
	currFileData = []
	for i in range(numSets):
		for j in range(numFolds):
			currFileData.append(sortedData[i+j*numSets])
			# print(i+j*numSets)
		# end for j
	# end for i


	# for each in currFileData:
	# 	print(each)
	# print(currFileData)
	AllFileData.append(currFileData)

# end for currFilename

print("Finished Reading all Files.")


```

```python
### Write Files out to new CSV


FinalOutput = []
for i in range(len(AllFileData[0])): # loop over all rows in each file
	# Stitch all files first/second/etc rows together
	currOutputRow = []
	for currFileData in AllFileData:
		currOutputRow.extend(currFileData[i][:])
		currOutputRow.extend([' '])
	# end for
	FinalOutput.append(currOutputRow)
# end of for i

with open(outputFilename, 'w', newline='') as csvfile:
	myWriter= csv.writer(csvfile, delimiter=',')
	for row in FinalOutput:
		myWriter.writerow(row)
	# end for row
# end with open()

print("Saved CSV.")
```
