```python

# EvalFlag; 1 = Point to Point eval, 2 = Point to Window (8 sec tol)
EvalFlag = 1

if EvalFlag == 1:
	currResults = ResultsF1_Point2Point
	EvalMethod = "Point to Point"
elif EvalFlag == 2:
	currResults = ResultsF1_Point2Window
	EvalMethod = "Point to Window"
#end if
	

AllAvgResults=[]
AllStdDevResults=[]
for datasetIdx in range(1,len(currResults)): #loop over datasets, ignore 0 idx
	AvgResults = []
	StdDevResults = []
	for ws_idx in range(len(currResults[datasetIdx])): # loop over window sizes
		AvgResults.append(np.mean(currResults[datasetIdx][ws_idx]))
		StdDevResults.append(np.std(currResults[datasetIdx][ws_idx]))
	# end of for ws_idx
	
	AllAvgResults.append(AvgResults)
	AllStdDevResults.append(StdDevResults)
# end of for dataset


for datasetIdx in range(1,len(currResults)): #loop over datasets, ignore 0 idx
	print("Average Values for F1 Score, {} Dataset".format(DatasetNames[datasetIdx]))
	print("{:5s} {:5s} {:5s} {:5s} {:5s}".format("5s", "10s", "15s", "20s", "25s")) 
	for ws_idx in range(len(currResults[datasetIdx])): # loop over window sizes
		print("{:<5.1f} ".format(AllAvgResults[datasetIdx-1][ws_idx]), end = '')
		# print(StdDevsResults)
	# end of for ws_idx
	print("\n\n")
# end of for dataset

```

```python
### CONSTANTS FOR PLOTTING

# plotcolors = ['b', 'r', 'k', 'tab:orange', 'tab:purple']

plotcolors=[ # Colors taken from ColorBrewer2
	'#984ea3', # Purple
	'#ff7f00', # Orange
	'#377eb8', # Blue
	'#e41a1c', # Red
	'#4daf4a', # Green
	'#000000' # Black as final option 
]

plotformats = [
	'o',
	'h',
	'>',
	'D',
	's',
]

plotmkrsize = [
	'6', # o
	'7', # h
	'7', # > 
	'5', # D
	'7', # s
]



CI_MULTIPLIER = 1 # Multiplier to use on the StdDev bars; 2 is 95% CI, 1.0 is 68% CI
```

```python
plt.figure('w', dpi=150)
plt.subplots(figsize=(4, 2.5)) 

legend_scores = []
legend_confint = []
legend_confint_edges = []

for i in range(len(currResults)-1): #loop over datasets, ignore 0 idx
	# x_data = np.array(WindowSizes) + 0.4*((i%3)-1) # Offset from each dataset
	x_data = np.array(WindowSizes) # no offset
	y_data = np.array(AllAvgResults[i])/100 # Convert to decimal instead of percent
	currStdDevs = np.array(AllStdDevResults[i])/100 # Convert to decimal instead of percent
	legend_item, = plt.plot(x_data, y_data, color=plotcolors[i], mec='k', marker=plotformats[i], ms=plotmkrsize[i], linestyle='-.')
	legend_scores.append(legend_item)

	# plt.errorbar(x_data, y_data, yerr=CI_MULTIPLIER * currStdDevs, color=plotcolors[i], fmt=plotformats[i], capsize=5, ecolor=plotcolors[i], elinewidth=1.5, alpha = 0.6)

	# ### PLOT CONFIDENCE INTERVAL
	ax = plt.gca()
	ci_upper = y_data+CI_MULTIPLIER*currStdDevs # 95% CI = 2 StdDevs away for 
	ci_lower = y_data-CI_MULTIPLIER*currStdDevs
	ci_lower[ci_lower<-0.1] = -0.1 # Truncate lower limit to keep lower axis close to zero  

	legend_item = ax.fill_between(x_data, ci_lower, ci_upper, facecolor=plotcolors[i], alpha=.3, edgecolor=plotcolors[i], linewidth=1)
	legend_confint.append(legend_item)
	legend_item = ax.fill_between(x_data, ci_lower, ci_upper, facecolor='none', alpha=.5, edgecolor=plotcolors[i], linewidth=2)
	legend_confint_edges.append(legend_item)

# end of for dataset

# Plot blue data on top for aesthetics (green was covering the data point at 10 sec)
i=2
x_data = np.array(WindowSizes) # no offset
y_data = np.array(AllAvgResults[i])/100 # Convert to decimal instead of percent
plt.plot(x_data, y_data, color=plotcolors[i], mec='k', marker=plotformats[i], ms=plotmkrsize[i], linestyle='-.')

# Plot Purple data on top for aesthetics (green was covering the data point at 20 sec)
i=0
x_data = np.array(WindowSizes) # no offset
y_data = np.array(AllAvgResults[i])/100 # Convert to decimal instead of percent
plt.plot(x_data, y_data, color=plotcolors[i], mec='k', marker=plotformats[i], ms=plotmkrsize[i], linestyle='-.')


# plt.title("{} F1 vs Window Size".format(EvalMethod)) 
plt.ylabel("F1 Score")
plt.xlabel("Window Size [sec]")

plt.grid(axis='y')
# plt.grid(axis='x')

plt.xlim([3,27])
plt.xticks(range(5,26,5))
# plt.xlim([3,32])
# plt.xticks(range(5,31,5))


LEGEND_HEIGHT = -0.4
lgd_1 = plt.legend(legend_scores, DatasetNames[1:], 
		   bbox_to_anchor=(-0.20, LEGEND_HEIGHT), loc='center left',
					ncol=3
		   # loc='outside lower center'
		  )
# plt.legend(DatasetNames[1:], location='bottom')


plt.savefig("Figures/F1vsWS_{}_CGv3_05to25.pdf".format(EvalMethod), bbox_extra_artists=(lgd_1,), bbox_inches='tight')
```
