```python

totalDomBites = 0
totalNonDomBites = 0
totalLeftBites = 0
totalRightBites = 0
totalBothBites = 0

# gt_handed_labels = gt_handed_labels[0:2]

for i in range(len(gt_handed_labels)):
	
	meal_gt_hands = gt_handed_labels[i]
	np_gt_hands = np.array(meal_gt_hands) # convert for logical broadcasting
	
	currDomHand = handedness[i]
	if currDomHand == 'right':
		DomHandNumber = 0
		NonDomHandNumber = 1
	else:
		DomHandNumber = 1
		NonDomHandNumber = 0
	
	totalDomBites += np.sum(np_gt_hands == DomHandNumber)
	totalNonDomBites += np.sum(np_gt_hands == NonDomHandNumber)
	totalRightBites += np.sum(np_gt_hands == 0)
	totalLeftBites += np.sum(np_gt_hands == 1)
	totalBothBites += np.sum(np_gt_hands == 2)
# end of for meal

print(gt_handed_labels[0])
print(gt_handed_labels[1])

print("Dom Bites    = {:-5d}".format(totalDomBites))
print("NonDom Bites = {:-5d}".format(totalNonDomBites))
print("Both Bites   = {:-5d}\n".format(totalBothBites))
print("Right Bites  = {:-5d}".format(totalRightBites))
print("Left Bites   = {:-5d}".format(totalLeftBites))
```
