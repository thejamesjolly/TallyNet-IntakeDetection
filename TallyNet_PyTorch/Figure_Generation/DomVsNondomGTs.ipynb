```python
biteHands = ClemDataset['bites_handedness_gt']
DomLabel = ClemDataset['handedness']

DomHand = 0
NonDomHand = 0
BothHand = 0

for i in range(len(biteHands)):
	for j in range(len(biteHands[i])):
		if biteHands[i][j] == 'right':
			DomHand += 1
		elif biteHands[i][j] == 'both':
			BothHand += 1
		else:
			NonDomHand +=1
	# end
# end

TotalBites = DomHand + BothHand + NonDomHand

print("Dom count    = {:8.2f}".format(DomHand/TotalBites*100))
print("NonDom count = {:8.2f}".format(NonDomHand/TotalBites*100))
print("Both count   = {:8.2f}".format(BothHand/TotalBites*100))
```
